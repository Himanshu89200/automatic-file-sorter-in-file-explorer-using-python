<h1>Automatic File storer in File explorer</h1>

```python
import os,shutil
```

```python
path = r"D:/downloads/"
```

```python
file_name = os.listdir(path)
```

```python
folder_names=['pdf files','image files','text files','word files']
for loop in range (0,4):
    if not os.path.exists(path + folder_names[loop]):
        print(path + folder_names[loop])
        os.makedirs(path + folder_names[loop])
```

```python
for file in file_name:
    if ".pdf" in file and not os.path.exists(path +"pdf files/" + file):
        shutil.move(path + file, path + "pdf files/" + file)
    elif ".jpg" in file and not os.path.exists(path +"image files/" + file):
        shutil.move(path + file, path + "image files/" + file)
    elif ".txt" in file and not os.path.exists(path +"text files/" + file):
        shutil.move(path + file, path + "text files/" + file)
    elif ".docx" in file and not os.path.exists(path +"word files/" + file):
        shutil.move(path + file, path + "word files/" + file)
   
```
